```python
import math
import cmath
import numpy as np
import matplotlib.pyplot as plt

def x(t):
  if t<0:
    return 0
  elif 0<=t and t<=1:
    return 1
  elif 1<t and t<=2:
    return (-t+2)
  elif 2<t:
    return 0

plt.figure(0)
for t in np.linspace(-3,3,500):
  plt.plot(t,x(t),'r')

plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t)')
plt.ylabel('x')
plt.xlabel('t')
plt.show()


plt.figure(1)
for t in np.linspace(-3,3,500):
  plt.plot(t,x(t+1),'r')

plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t+1)')
plt.ylabel('x')
plt.xlabel('t')
plt.show()


plt.figure(2)
for t in np.linspace(-3,3,500):
  plt.plot(t,x(-t+1),'r')
  #print(f"x({t})={x(t)}\n")

plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(-t+1)')
plt.ylabel('x')
plt.xlabel('t')
plt.show()

plt.figure(3)
for t in np.linspace(-3,3,500):
  plt.plot(t,x(3/2*t),'r')
  

plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(3t/2)')
plt.ylabel('x')
plt.xlabel('t')
plt.show()

plt.figure(4)
for t in np.linspace(-3,3,500):
  plt.plot(t,x(3/2*t+1),'r')
  

plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(3t/2+1)')
plt.ylabel('x')
plt.xlabel('t')
plt.show()


```

```python
import math
import cmath
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  if 0>x:
    return 0
  elif 0<=x and x<=1:
    return 1
  elif 1<x and x<=2:
    return (-x+2)
  elif 2<x:
    return 0
  

x = np.linspace(-3,3,100)

y =  np.vectorize(f,otypes=[float])(x)


plt.figure(0)
plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t)')
plt.ylabel('x')
plt.xlabel('t')

plt.plot(x,y, 'r')
plt.show()


plt.figure(0)
plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t)')
plt.ylabel('x')
plt.xlabel('t')

# plot the function
plt.plot(x+1,y, 'r')

# show the plot
plt.show()

plt.figure(0)
plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t)')
plt.ylabel('x')
plt.xlabel('t')

# plot the function
plt.plot(-x+1,y, 'r')

# show the plot
plt.show()

plt.show()

plt.figure(0)
plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t)')
plt.ylabel('x')
plt.xlabel('t')

# plot the function
plt.plot(3/2*x,y, 'r')

# show the plot
plt.show()


plt.show()

plt.figure(0)
plt.axhline(y=0)
plt.axvline(x=0)
plt.title('x(t)')
plt.ylabel('x')
plt.xlabel('t')

# plot the function
plt.plot(3/2*x+1,y, 'r')

# show the plot
plt.show()
```
