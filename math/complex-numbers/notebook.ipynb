```python
from math import pi, cos, sin
import matplotlib.pyplot as plt

n = int(input("Ingrese el numero de n: "))
z = list()
for k in range(0,n):
  zaux = complex(cos(2*pi*k/n) , sin(2*pi*k/n))
  z.append(zaux)

#print(f"z={z}")

plt.figure(0)
for k in range(0,n):
  plt.plot(z[k].real,z[k].imag,'ro')
  
plt.axhline(y=0)
plt.axvline(x=0)
plt.title('Plano z')
plt.ylabel('Imag')
plt.xlabel('Real')
plt.show()
```
